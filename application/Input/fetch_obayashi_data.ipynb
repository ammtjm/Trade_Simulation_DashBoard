```python
from pathlib import Path
from datetime import date, timedelta

from fetch_obayashi_data import run

# 直近5年分を取得
end_date = date.today()
start_date = end_date - timedelta(days=365 * 5)

paths = run(
    ticker="1802.T",
    start=start_date,
    end=end_date,
    output_dir=Path("../Output"),
    news_query="大林組",
    news_limit=20,
)
paths
```
